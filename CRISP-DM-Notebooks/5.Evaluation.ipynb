# 5. Modelling

## Results

The `SVM` model successfully met the business criteria. As per business criteria we concluded that:
- All digits are predicted with 80% accuracy or more (general accuracy being 93%)
- General target accuracy of 90% is achieved.
- Cross validation proved that the model is not overfitting and can generalize well.

## Reviewing the Process

In general, the process had 2 major drawbacks:

1. **Time**: The code execution was performed on a laptop, thus often speed was prioritized over accuracy (for example, at hyperparameter tuning). This was done because of the limited time (5 weeks) available for this project, and because the acquisition of results was more important than the accuracy of the model.
2. **Iterations**: This project included only 1 iteration. This was due to the time constraints and because of the heavier emphases on researched rather than experimentation. Thus, not all possibilities were explored. Not all models were tested, and not all hyperparameters were tuned and only one type of feature cleaning phase was performed. If different combinations of steps would have been tested, the results could have been different.

The following steps were fully completed:
- Feature extraction: a lot of automatic and manual features were extracted.
- Model validation: Cross validation was performed to ensure that the model is not overfitting. Furthermore, each fold confirmed that the model is not overfitting and can generalize well. The confusion matrix also showed, that in general, the model can indeed predict all labels well.

The following steps could be improved:
- Model hyperparameter tuning: Not all hyperparameters were tuned, and more combinations could have been tested.
- Testing different models other than SVM, such as Random Forest, Decision Trees, etc.
- Image preprocessing: The stroke with of the images was not standardized.
- Research: More research could have been done on the topic of image preprocessing and feature extraction. This was mainly due to time constraints.

## Next Steps

The following next steps should be taken in order to improve the model:
1. Conduct more research on best practices.
2. Standardize the stroke width of the images.
3. Trying out different ways and combinations of data cleaning: remove more or less outliers, being more generic or more specific with outliers per category
4. `RandomForest` could be explored next, as per research recommendation.
5. Multiple iterations: check which combination of preprocessing, feature extraction, and model works best together.
